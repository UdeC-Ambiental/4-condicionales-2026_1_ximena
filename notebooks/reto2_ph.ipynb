# 🧩 Reto 2 – pH
**Objetivo:** clasificar pH y generar alerta química.

**Datos:** `../data/muestras_calidad_agua.csv`

```python
# Escribe tu código aquí
import pandas as pd

df = pd.read_csv("../data/muestras_calidad_agua.csv")
df.head()
```

```python
for index, fila in df.iterrows():
    print(fila)
    
```

```python
for index, fila in df.iterrows():

    sitio = fila["sitio"]
    ph = fila["pH"]

    # Clasificación del agua
    if ph < 6.5:
        estado = "Ácida"

    elif ph <= 8.5:
        estado = "Normal"

    else:
        estado = "Alcalina"

    # Alertas químicas
    if ph < 5:
        alerta = "⚠️ Riesgo químico alto"

    elif ph <= 8.5:
        alerta = " Agua segura"

    else:
        alerta = "⚠️ Riesgo alcalino"

    print(f"Sitio: {sitio}")
    print(f"pH: {ph}")
    print(f"Estado: {estado}")
    print(f"Alerta: {alerta}")
    print("---------------------------")
```

```python
# Filtrar muestras con problemas de pH
df_alerta = df[(df["pH"] < 6.5) | (df["pH"] > 8.5)]

# Mostrar resultados filtrados
print(df_alerta)

# Exportar a CSV
df_alerta.to_csv("../data/resultado_ph.csv", index=False)

print("Archivo resultado_ph.csv exportado correctamente")
```

Promedio y desviacion estandar

```python
# Promedio del pH
promedio_ph = df["pH"].mean()

# Desviación estándar del pH
desviacion_ph = df["pH"].std()

print(f"Promedio de pH: {promedio_ph}")
print(f"Desviación estándar del pH: {desviacion_ph}")
```
