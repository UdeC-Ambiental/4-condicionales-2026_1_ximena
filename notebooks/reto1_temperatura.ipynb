```python
import pandas as pd

# Leer archivo CSV
df = pd.read_csv("../data/muestras_calidad_agua.csv")

# Función
def riesgo_temperatura(temp):

    if temp < 20:
        return "Baja"

    elif temp <= 26:
        return "Media"

    else:
        return "Alta"

# Crear nueva columna
df["riesgo_termico"] = df["temperatura"].apply(riesgo_temperatura)

# Imprimir resultados
print(df[["sitio", "temperatura", "riesgo_termico"]])
```

```python
import pandas as pd

# Leer CSV
df = pd.read_csv("../data/muestras_calidad_agua.csv")

# Función de clasificación
def riesgo_temperatura(temp):

    if temp < 20:
        return "Baja"

    elif temp <= 26:
        return "Media"

    else:
        return "Alta"

# Crear nueva columna
df["riesgo_termico"] = df["temperatura"].apply(riesgo_temperatura)

# Mostrar resultados
print(df[["sitio", "temperatura", "riesgo_termico"]])

# Exportar nuevo CSV
df.to_csv("../data/resultados_temperatura.csv", index=False)

print("Archivo CSV generado correctamente")
```
