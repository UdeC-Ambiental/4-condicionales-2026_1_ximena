```python
import pandas as pd
```

```python
datos = pd.read_csv("../data/muestras_calidad_agua.csv")
```

```python
datos.head()
```

```python
# Información general del DataFrame
datos.info()
```

```python
# Estadísticas descriptivas
datos.describe()
```

```python
for columna in datos.columns:
    print(columna)
```

```python
# =========================================
# Cantidad de filas y columnas
# =========================================

print("Dimensiones del DataFrame:")
print(datos.shape)
```

```python
# =========================================
# Valores nulos por columna
# =========================================

tabla_nulos = datos.isnull().sum().reset_index()

tabla_nulos.columns = ["Columna", "Valores nulos"]

tabla_nulos
```

```python
# =========================================
# Estadísticas descriptivas más organizadas
# =========================================

estadisticas = datos.describe()

estadisticas
```

```python
# =========================================
# Promedio de variables numéricas
# =========================================

promedios = datos.mean(numeric_only=True).reset_index()

promedios.columns = ["Variable", "Promedio"]

promedios
```

```python
# =========================================
# Valores mínimos
# =========================================

minimos = datos.min(numeric_only=True).reset_index()

minimos.columns = ["Variable", "Valor mínimo"]

minimos
```

```python
# =========================================
# Valores máximos
# =========================================

maximos = datos.max(numeric_only=True).reset_index()

maximos.columns = ["Variable", "Valor máximo"]

maximos
```

```python
# =========================================
# Evaluación del pH promedio
# =========================================

promedio_ph = datos["pH"].mean()

print(f"Promedio de pH: {promedio_ph:.2f}")

if promedio_ph > 7:
    print("El agua tiende a ser básica")
elif promedio_ph == 7:
    print("El agua es neutra")
else:
    print("El agua tiende a ser ácida")
```

```python
# =========================================
# Evaluación de temperatura
# =========================================

promedio_temperatura = datos["temperatura"].mean()

print(f"Promedio de temperatura: {promedio_temperatura:.2f} °C")

if promedio_temperatura > 25:
    print("La temperatura del agua es alta")
else:
    print("La temperatura del agua es normal")
```
